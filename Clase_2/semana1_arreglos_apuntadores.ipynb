Se aplica el condicional if/else, si el elemento evaluado es igual al buscado, se muestra el mensaje: “El número buscado está en la lista” y se detiene el ciclo; si no esta el numero, el programa no hará nada y continuará con la búsqueda silenciosamente.

### Ejercicio 8
Implementar búsqueda binaria en lista ordenada.

<details><summary>Click here for the solution</summary>

```python
# Casos de prueba y solución
lista = [1,2,3,4,5]
obj = 4
low, high = 0, len(lista)-1
encontrado = -1
while low <= high:
    mid = (low+high)//2
    if lista[mid] == obj:
        encontrado = mid
        break
    elif lista[mid] < obj:
        low = mid+1
    else:
        high = mid-1
print(encontrado)
```

### Ejercicio 9
Eliminar duplicados manteniendo el orden.

````python
lista=[2,5,9,3,2,7,9,9]
noDuplicados=[]
for i in lista:
    if i not in noDuplicados:
        noDuplicados.append(i)
    else: pass
print(f"La lista {lista} sin duplicados es: {noDuplicados}")
````

Se crean dos listas, una con elementos (lista) y una vacía (noDuplicados), después se inicia un ciclo for, que se repetirá el número de elementos que tenga lista. Si el número evaluado no se encuentra en la lista noDuplicados entonces lo insertará al final de la lista, si el número ya está en la lista de noDuplicados (gracias a ciclos anteriores), entonces no hará nada con el número. Al final mostrará la lista original y la que contiene los elementos sin duplicados. 

<details><summary>Click here for the solution</summary>

```python
# Casos de prueba y solución
lista = [1,2,2,3]
resultado = []
for x in lista:
    if x not in resultado:
        resultado.append(x)
print(resultado)
```

### Ejercicio 10
Intercalar dos listas en una nueva lista.

````python
lista1=[1,1,1,1]
lista2=[2,3,4,5]
nuevaLista=lista1[:]
posicion = 1
for i in lista2:
    nuevaLista.insert(posicion, i)
    posicion = posicion + 2
print("La lista 1 y la lista 2 intercaladas son la siguiente lista: ", nuevaLista)
````

Se inicia un for que se repetirá el número de elementos que hay en lista2. En este ciclo se inserta el elemento en la posición uno de la lista y luego se le suma 2 a la posición para que en el siguiente ciclo el elemento se inserta en la posición 3 (y se intercalen los valores de la lista 1 y 2). Al final el programa muestra la nueva lista. 

<details><summary>Click here for the solution</summary>

