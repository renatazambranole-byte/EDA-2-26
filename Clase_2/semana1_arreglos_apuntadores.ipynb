### Ejercicio 3
Encontrar el elemento máximo y mínimo sin usar `max()` ni `min()`.

<details><summary>Click here for the solution</summary>

```python
# Casos de prueba y solución
lista = [3,1,4]
minimo = lista[0]
maximo = lista[0]
for x in lista:
    if x < minimo: minimo = x
    if x > maximo: maximo = x
print(minimo, maximo)
```

### Ejercicio 8
Implementar búsqueda binaria en lista ordenada.

<details><summary>Click here for the solution</summary>

```python
# Casos de prueba y solución
lista = [1,2,3,4,5]
obj = 4
low, high = 0, len(lista)-1
encontrado = -1
while low <= high:
    mid = (low+high)//2
    if lista[mid] == obj:
        encontrado = mid
        break
    elif lista[mid] < obj:
        low = mid+1
    else:
        high = mid-1
print(encontrado)
```

</details>

### Ejercicio 10
Intercalar dos listas en una nueva lista.

<details><summary>Click here for the solution</summary>

```python
# Casos de prueba y solución
a = [1,3,5]
b = [2,4,6]
resultado = []
for x,y in zip(a,b):
    resultado.extend([x,y])
print(resultado)
```

### Ejercicio 13
Implementar una función que intercambie dos elementos de una lista por referencia.

<details><summary>Click here for the solution</summary>

```python
# Casos de prueba y solución
def swap(lista, i, j):
    lista[i], lista[j] = lista[j], lista[i]
lista = [1,2,3]
swap(lista,0,2)
print(lista)
```
