Para esta asignación se recomienda que utilices la siguiente implemetnación para el arreglo pilas.

```python
class Empty(Exception):
    """Error attempting to access an element from an empty container."""
    pass

class ArrayStack:
    """LIFO Stack implementation using a Python list as underlying storage."""

    def __init__(self):
        """Create an empty stack."""
        self._data = []  # nonpublic list instance

    def __len__(self):
        """Return the number of elements in the stack."""
        return len(self._data)

    def is_empty(self):
        """Return True if the stack is empty."""
        return len(self._data) == 0

    def push(self, e):
        """Add element e to the top of the stack."""
        self._data.append(e)  # new item stored at end of list

    def top(self):
        """Return (but do not remove) the element at the top of the stack.
        Raise Empty exception if the stack is empty.
        """
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data[-1]  # the last item in the list

    def pop(self):
        """Remove and return the element from the top of the stack (i.e., LIFO).
        Raise Empty exception if the stack is empty.
        """
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data.pop()  # remove last
    
    def remove(self):
        if self.is_empty():
            print("stack is empty")
            return self
        else:
            self.pop()
            return self.remove()
    def mostrar(self):
        print(self._data)
```

**R-6.3** Implement a function with signature transfer(S, T) that transfers all elements from stack $S$ onto stack $T$, so that the element that starts at the top of $S$ is the first to be inserted onto $T$, and the element at the bottom of $S$ ends up at the top of $T$.

```python
def transfer(S, T):
    largo = len(S)
    for i in range(largo):
        T.push(S.pop())
    print("Esta es la lista T: ")
    T.mostrar()
S= ArrayStack()
S.push(1)
S.push(2)
S.push(3)
S.push(4)
S.push(5)
T= ArrayStack()
print("Esta es la lista S: ")
S.mostrar()
transfer(S,T)
```

Se crea la función transfer que recibe dos objetos de tipo ArrayStack, S y T; guarda en la variable largo la extensión de S e inicia un ciclo for que va desde 0 hasta largo. Dentro del ciclo se saca el último elemento de S con pop() y se inserta directamente en T usando push(), pasando los elementos uno por uno. Afuera se crean los objetos S y T, se le añaden los valores a S, se muestra la lista original y se llama a la función para imprimir la lista T con los elementos invertidos.

**R-6.5** Implement a function that reverses a list of elements by pushing them onto a stack in one order, and writing them back to the list in reversed order.

```python
def reverse(A):
    R = ArrayStack()  
    for i in A:
        R.push(i)
    for i in range(len(A)):
        A[i] = R.pop()
    print("Lista después de la función: ", A)
A=[2,4,2,4]
print("Lista original: ",A)
reverse(A)
```

La función reverse toma una lista y crea un objeto de tipo ArrayStack. Después inicia un ciclo for donde mete los elementos de A en R usando push. Después, otro ciclo for con rango del largo de A lo que hará será colocar los elementos de R uno por uno en A nuevamente, pero ahora quedan revertidos por cómo se metieron en R (mientras la función se desarrolla,se usa pop para vaciar R). Las últimas líneas llaman a la función y muestran la lista original.

**C-6.16** Modify the ArrayStack implementation so that the stack’s capacity is limited to maxlen elements, where maxlen is an optional parameter to the constructor (that defaults to None). If push is called when the stack is at full capacity, throw a Full exception (defined similarly to Empty).

```python
class Empty(Exception):
    """Error attempting to access an element from an empty container."""
    pass
class Full(Exception):
    pass

class ArrayStack:
    """LIFO Stack implementation using a Python list as underlying storage."""

    def __init__(self, maxlen=None):
        """Create an empty stack."""
        self._data = []  # nonpublic list instance
        self.maxlen = maxlen

    def __len__(self):
        """Return the number of elements in the stack."""
        return len(self._data)

    def is_empty(self):
        """Return True if the stack is empty."""
        return len(self._data) == 0

    def push(self, e):
        """Add element e to the top of the stack."""
        if self.maxlen is not None and len(self._data) == self.maxlen:
            raise Full("Stack está lleno")
        else:
            self._data.append(e)  # new item stored at end of list

    def top(self):
        """Return (but do not remove) the element at the top of the stack.
        Raise Empty exception if the stack is empty.
        """
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data[-1]  # the last item in the list

    def pop(self):
        """Remove and return the element from the top of the stack (i.e., LIFO).
        Raise Empty exception if the stack is empty.
        """
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data.pop()  # remove last
    
    def remove(self):
        if self.is_empty():
            print("stack is empty")
            return self
        else:
            self.pop()
            return self.remove()
    def mostrar(self):
        print(self._data)
stack = ArrayStack(maxlen=2)
stack.push(1)
stack.push(2)
stack.mostrar()
try:
    print("Intentando añadir un elemento fuera del límite")
    stack.push(3)
except Full as error:
    print(f"Error econtrado: {error}")
```

Primero se crea la excepcion Full para gestionar los errores de la pila. Después se define la clase ArrayStack agregando el atributo opcional self.maxlen en el constructor para limitar su tamaño; en el método push() se valida que si self.maxlen no es nulo y la cantidad de elementos en self._data alcanza ese límite, se lanza la excepción Full. Al final se instancia stack con un límite de 2 elementos, se le ingresan dos datos con push() para después mostrarlos con mostrar(), y finalmente, dentro de un bloque try-except, se intenta meter un tercer valor para capturar el error sin que se detenga el programa y mostrar el mensaje en terminal.

**C-6.17** In the previous exercise, we assume that the underlying list is initially empty. Redo that exercise, this time preallocating an underlying list with length equal to the stack’s maximum capacity.

```python
class Empty(Exception):
    pass

class Full(Exception):
    pass

class ArrayStack:
    def __init__(self, maxlen=None):
        self.maxlen = maxlen
        self._size = 0  
        if maxlen is not None:
            self._data = [None] * maxlen
        else:
            self._data = []

    def __len__(self):
        return self._size

    def is_empty(self):
        return self._size == 0

    def push(self, e):
        if self.maxlen is not None:
            if self._size == self.maxlen:
                raise Full("Stack está lleno")
            self._data[self._size] = e
            self._size += 1
        else:
            self._data.append(e)
            self._size += 1

    def top(self):
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data[self._size - 1]

    def pop(self):
        if self.is_empty():
            raise Empty('Stack is empty')
        self._size -= 1
        val = self._data[self._size]
        self._data[self._size] = None  
        return val

    def mostrar(self):
        print(self._data[:self._size])
try:
    stack=ArrayStack(maxlen=2)
    stack.push(1)
    stack.push(2)
    stack.mostrar()
    print("Intentando añadir un elemento fuera del límite:")
    stack.push(3)
except Full as error:
    print(f"Error encontrado: {error}")
```

Se cambió el constructor para preasignar la lista self._data con un tamaño fijo de posiciones None según el valor de maxlen, y se añadió la variable self._size como contador manual de elementos. En el método push() se eliminó el uso de append() para guardar los datos sobrescribiendo directamente la posición del índice self._size. Por último, en el método pop() se quitó el pop() de Python para en su lugar restar 1 al contador, limpiar el casillero asignándole None nuevamente y devolver el valor recuperado.
