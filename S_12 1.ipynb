**R-14.14** A simple undirected graph is complete if it contains an edge between every pair of distinct vertices. What does a depth-first search tree of a complete graph look like?

El árbol de búsqueda en profundidad de un grafo completo es un camino, es decir, una cadena en la que cada vértice tiene un solo hijo. Esto pasa porque en un grafo completo cada vértice está conectado con todos los demás. Cuando la búsqueda empieza en un vértice, siempre hay algún vecino sin visitar, así que se mueve a él y desde ahí repite lo mismo. Nunca se queda sin opciones hasta haber visitado todos los vértices, por lo que no se forman ramas. Si el grafo tiene n vértices, el árbol queda con n − 1 aristas y altura n − 1. Todas las demás aristas del grafo, que son la gran mayoría, quedan como aristas de retroceso que van de un vértice a alguno de sus ancestros en la cadena.

**R-14.15** Recalling the definition of a complete graph from Exercise R-14.14, what does a breadth-first search tree of a complete graph look like?

El árbol de búsqueda en anchura de un grafo completo es una estrella, la raíz queda conectada directamente con todos los demás vértices, y cada uno de ellos es una hoja. Esto sucede porque en un grafo completo el vértice donde empieza la búsqueda es vecino de todos los otros, así que los descubre a todos en el primer nivel y no queda ningún vértice para un segundo nivel. Si el grafo tiene n vértices, la raíz tiene n − 1 hijos y el árbol tiene altura 1. Las demás aristas del grafo, es decir, las que conectan a los hijos entre sí, no forman parte del árbol y quedan como aristas de cruce entre vértices del mismo nivel.

Es lo contrario de lo que pasa en la búsqueda en profundidad del ejercicio anterior, donde el árbol era un camino largo con altura n − 1. Con el mismo grafo, una búsqueda produce el árbol más alto posible y la otra el más bajo.

**R-14.24** Show how to modify the pseudo-code for Dijkstra’s algorithm for the case when the graph is directed and we want to compute shortest directed paths from the source vertex to all the other vertices.

Para un grafo dirigido, el algoritmo de Dijkstra se queda casi igual, y el único cambio está en el paso donde se actualizan las distancias. En un grafo no dirigido, cuando sacamos de la cola el vértice v con menor distancia, revisamos todos los vértices adyacentes a v. En uno dirigido solo debemos revisar las aristas que salen de v, es decir, los vértices a los que se puede llegar siguiendo la dirección de la arista, porque las que llegan a v no sirven para ir desde el origen hacia otros vértices. Todo lo demás se mantiene, la distancia del origen empieza en 0, las de los demás vértices en infinito, y en cada vuelta se saca de la cola el vértice con menor distancia y se relajan las aristas que salen de él, actualizando la distancia de un vecino z si pasar por v resulta más corto. Al terminar, la distancia de cada vértice es la longitud del camino dirigido más corto desde el origen, y si algún vértice no se puede alcanzar siguiendo las direcciones de las aristas, su distancia se queda en infinito.


1. Poner la distancia del Origen en 0.
2. Poner la distancia de todos los demás vértices en Infinito.
3. Agrega todos los vértices a una lista.

4. Mientras la lista no esté vacía:
    a. Elige el vértice de la lista que tenga la menor distancia y llámalo V.
    b. Sacarlo de la lista.
    c. Revisar solo las aristas con flecha que salen de V hacia un vecino Z:
        - Si (distancia de V + peso de la arista) es menor que la distancia de Z:
            - Cambia la distancia de Z por esa nueva distancia más corta.

5. Devuelve las distancias finales de todos los vértices.

**R-14.28** Describe the meaning of the graphical conventions used in Figure 14.9 illustrating a DFS traversal. What do the line thicknesses signify? What do the arrows signify? How about dashed lines?

En la Figura 14.9, el grosor de las líneas indica en qué estado está cada conexión durante el recorrido. Las líneas delgadas son las aristas que todavía no se han explorado, y la línea gruesa marca la arista que el algoritmo está revisando en ese momento. Las flechas muestran la dirección en la que avanza la búsqueda, es decir, hacia qué vértice se mueve al recorrer una arista. Las líneas punteadas representan las aristas de retroceso (back edges), que son las que llevan a un vértice que ya se había visitado antes y por eso no forman parte del árbol. En cambio, las líneas sólidas son las aristas de descubrimiento (tree edges), que son las que permiten llegar a vértices nuevos y van formando el árbol de la búsqueda en profundidad.

**C-14.45** Say that a maze is constructed correctly if there is one path from the start to the finish, the entire maze is reachable from the start, and there are no loops around any portions of the maze. Given a maze drawn in an $n \times n$ grid, how can we determine if it is constructed correctly? What is the running time of this algorithm?e in the same connected component.

Para saber si un laberinto está bien construido, lo primero es verlo como un grafo, cada celda de la cuadrícula es un vértice, y dos celdas vecinas están conectadas por una arista cuando no hay una pared entre ellas. Después se hace una búsqueda en profundidad a partir de la entrada. Si al terminar se visitaron todas las celdas, el laberinto es conexo, es decir, todo se puede alcanzar desde el inicio, y eso incluye la salida. Además, durante la búsqueda hay que fijarse en si aparece alguna arista de retroceso, que es la que lleva a una celda ya visitada que no es de donde venimos. Si aparece una, hay un ciclo, y por lo tanto un lazo en el laberinto, así que no está bien construido. Si se visitan todas las celdas y no aparece ninguna arista de retroceso, entre la entrada y la salida hay un solo camino, entonces el laberinto está bien construido. El tiempo de ejecución es O(n²), porque la búsqueda en profundidad toma O(V + E), la cuadrícula tiene n² vértices y cada celda se conecta como máximo con cuatro vecinas, así que el número de aristas también es del orden de n².
