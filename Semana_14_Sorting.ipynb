**R-12.7** Suppose we are given two $n$-element sorted sequences $A$ and $B$ each with distinct elements, but potentially some elements that are in both sequences. Implement an $O(n)$-time method for computing a sequence representing the union $A \cup B$ (with no duplicates) as a sorted sequence.

```python
class Union:
    def __init__(self):
        pass
    def AUnionB(self, A, B):
        AB = []
        i = 0
        j = 0
        while i < len(A) and j < len(B):
            if A[i] < B[j]:
                AB.append(A[i])
                i += 1
            elif A[i] > B[j]:
                AB.append(B[j])
                j += 1
            else:
                AB.append(A[i])
                i += 1
                j += 1
        while i < len(A):
            AB.append(A[i])
            i += 1

        while j < len(B):
            AB.append(B[j])
            j += 1
        return AB
```

What happens in the partition step when there are elements equal to the pivot? What is the running time of the algorithm if all the input elements are equal?

Para demostrar que el algoritmo ordena bien con elementos repetidos, se usa inducción (demostrar que algo se cumple para todos los casos probando que funciona en el caso base y que cada paso garantiza el siguiente.) sobre el tamaño de la secuencia. Con 0 o 1 elementos ya está ordenada. Para una secuencia más grande, después de la partición el pivote queda en su posición final, lo de su izquierda es menor o igual y lo de su derecha es mayor o igual. Esto sigue siendo cierto con repetidos, porque un elemento igual al pivote puede ir de cualquier lado sin romper el orden. Como las dos partes son más pequeñas que la original, por hipótesis de inducción quedan ordenadas, y con el pivote en medio forman la secuencia completa ordenada.

En el paso de partición, los índices se detienen cuando encuentran un elemento igual al pivote, así que esos elementos se intercambian y se reparten entre las dos partes en lugar de irse todos a un solo lado. Por eso, si todos los elementos son iguales, el pivote termina en la mitad y las dos partes quedan casi del mismo tamaño. Cada nivel de la recursión recorre los n elementos y hay alrededor de log n niveles, así que el tiempo de ejecución es O(n log n).


**R-12.18** Describe a radix-sort method for lexicographically sorting a sequence $S$ of triplets $(k, l, m)$, where $k, l$, and $m$ are integers in the range $[0, N - 1]$, for some $N \geq 2$. How could this scheme be extended to sequences of $d$-tuples $(k_1, k_2, \ldots, k_d)$, where each $k_i$ is an integer in the range $[0, N - 1]$?

Para ordenar lexicográficamente una secuencia de tripletas (k, l, m), se usa radix sort, que consiste en ordenar la secuencia varias veces, una por cada componente de la tripleta, empezando por la última. Primero se ordenan todas las tripletas según m, luego según l y al final según k. Cada una de estas pasadas se hace con un ordenamiento estable, que como los números están en el rango de 0 a N − 1, tarda O(n + N). Que el ordenamiento sea estable es lo que hace que funcione, porque quiere decir que si dos elementos tienen el mismo valor en la componente que se está ordenando, se quedan en el mismo orden relativo en el que venían. Así, cuando se ordena por l, las tripletas con el mismo l conservan el orden que ya tenían por m, y cuando se ordena por k, las que tienen el mismo k conservan el orden de l y m. Al terminar, la secuencia queda ordenada primero por k, después por l y después por m, que es el orden lexicográfico.

Esto se puede extender de forma directa a d-tuplas (k1, k2, …, kd). Se hacen d pasadas del mismo ordenamiento estable, empezando por la última componente, kd, y terminando con la primera, k1. Gracias a la estabilidad, cada pasada respeta el orden que dejaron las anteriores. Como cada pasada tarda O(n + N), con n elementos en la secuencia, el tiempo total para las d-tuplas es O(d(n + N)), y para el caso de las tripletas, donde d es 3, queda en O(n + N).


**R-12.21** Given a sequence $S$ of $n$ values, each equal to 0 or 1, describe an in-place method for sorting $S$.

```python
def sort(S):
    i = 0
    j = len(S) - 1

    while i < j:
        if S[i] == 0:
            i += 1
        elif S[j] == 1:
            j -= 1
        else:
            temp = S[i]
            S[i] = S[j]
            S[j] = temp
            i += 1
            j -= 1
```

```python
S = [1,0,1,0,1,0,0,0,1,1,0,1,1,0]
print("S desordenada: ", S)
sort(S)
print("S ordenada: ", S)
```

Para ordenar la secuencia S sin crear otra lista, se usan dos índices, i y j. El índice i empieza al inicio de la secuencia, en la posición 0, y j empieza al final, en la última posición. Mientras i sea menor que j, se van revisando los elementos desde los extremos hacia el centro. Si S[i] ya es un 0, está en el lado correcto, así que solo se avanza i. Si S[j] ya es un 1, también está bien ubicado, así que se retrocede j. Cuando ninguno de los dos casos se cumple, quiere decir que hay un 1 del lado izquierdo y un 0 del lado derecho, por lo que se intercambian de lugar usando una variable temporal y después se mueven ambos índices un paso hacia adentro. El proceso termina cuando los índices se cruzan, y para ese momento todos los ceros quedaron a la izquierda y todos los unos a la derecha. Como solo se recorre la secuencia una vez y no se usa otra lista, el tiempo de ejecución es O(n).

**R-12.23** What is the best algorithm for sorting each of the following: general comparable objects, long character strings, 32-bit integers, double-precision floating-point numbers, and bytes? Justify your answer.

Objetos comparables en general: el mejor algoritmo es merge sort o quick sort, porque solo se puede comparar un elemento con otro y ningún algoritmo que se base en comparaciones puede ser más rápido que O(n log n). Merge sort siempre tarda O(n log n), y quick sort en promedio también, y además se puede hacer sin usar otra lista.

Cadenas de caracteres largas: el mejor es radix sort. Comparar dos cadenas largas es costoso, porque hay que revisar carácter por carácter, y con un algoritmo de comparaciones ese costo se repite muchas veces. Con radix sort se ordena una pasada por cada posición de la cadena, empezando por la última, y cada pasada es rápida porque los caracteres tienen pocos valores posibles.

32 bit enteros: el mejor es radix sort. Un entero de 32 bits se puede ver como una tupla de 4 bytes, así que bastan 4 pasadas, y como cada pasada tarda O(n + N) con N = 256, en total tarda O(n), que es mejor que O(n log n).

Números de punto flotante de doble precisión: el mejor también es radix sort, porque cada número ocupa 64 bits y se pueden tratar como una tupla de 8 bytes, con 8 pasadas. Solo hay que hacer un pequeño ajuste con los números negativos, para que queden en el orden correcto. Si no se quiere hacer ese ajuste, quick sort también es una buena opción.

Bytes: el mejor es radix sort con una sola pasada. Un byte solo puede tener 256 valores distintos, así que el rango es pequeño y se ordena en tiempo O(n), sin necesidad de comparar elementos.


**C-12.25** Linda claims to have an algorithm that takes an input sequence $S$ and produces an output sequence $T$ that is a sorting of the $n$ elements in $S$.

a. Give an algorithm, is_sorted, that tests in $O(n)$ time if $T$ is sorted.

b. Explain why the algorithm is_sorted is not sufficient to prove a particular output $T$ to Linda’s algorithm is a sorting of $S$.

c. Describe what additional information Linda’s algorithm could output so that her algorithm’s correctness could be established on any given $S$ and $T$ in $O(n)$ time.

a)

```python
def is_sorted(T):
    for i in range(1, len(T)):
        if T[i-1] > T[i]:
            return False
    return True

print("Lista [1, 2, 3], is sorted? ", is_sorted([1,2,3]))
print("Lista [1, 3, 2], is sorted? ",is_sorted([1,3,2]))
```

b) La función is_sorted(T) no es suficiente, porque solo revisa que los elementos de T estén en orden creciente, pero no comprueba que T tenga exactamente los mismos elementos que la secuencia original S, ni que se repitan la misma cantidad de veces. Por ejemplo, si S = [5, 2, 8], el resultado correcto sería T = [2, 5, 8]. Sin embargo, el algoritmo de Linda podría devolver T = [1, 2, 3] o T = [2, 2, 2], y en los dos casos is_sorted(T) devolvería True, aunque T no sea un ordenamiento válido de S, porque no tiene los mismos elementos.

c) Para poder comprobar que el resultado es correcto en tiempo O(n), el algoritmo de Linda tendría que devolver, además de la secuencia T, un arreglo P de tamaño n que indique en qué posición de S estaba originalmente cada elemento de T, es decir, que cumpla T[i] = S[P[i]]. Con esa información extra solo hace falta recorrer los arreglos una vez para comprobar tres cosas. Primero, que T esté ordenada. Segundo, que todos los índices de P sean válidos y no se repitan, para lo cual se puede usar un arreglo auxiliar que marque los índices ya vistos. Y tercero, que cada elemento de T sea igual al elemento de S al que apunta P. Con esto se asegura que se usaron todos los elementos de S exactamente una vez, y como cada comprobación recorre n elementos, el tiempo total sigue siendo O(n).


**C-12.26** Describe and analyze an efficient method for removing all duplicates from a collection $A$ of $n$ elements.

```python
def noDuplicados(A):
    A.sort()
    B = [A[0]]
    i = 0
    for i in range (1,len(A)):
        if A[i] != B[-1]:
            B.append(A[i])
            i += 1
    return B
```

```python
A =[1,24,15,5,2,7,25,9,24,2,7,15,5]
print("Lista original: ", A)
print("Lista sin duplicados: ", noDuplicados(A))
```

Para quitar los duplicados de una colección A de n elementos, primero se ordena A, y así los elementos repetidos quedan uno junto a otro. Después se crea una lista nueva B con el primer elemento de A y se recorre el resto de A una sola vez, comparando cada elemento con el último que se agregó a B. Si son distintos, se agrega a B, y si son iguales, se ignora porque es un duplicado. Ordenar A tarda O(n log n) y el recorrido tarda O(n), así que el tiempo total es O(n log n), que es mejor que comparar cada elemento con todos los demás, que tardaría O(n²).

**C-12.27** Augment the PositionalList class (see Section 7.4) to support a method named merge with the following behavior. If $A$ and $B$ are PositionalList instances whose elements are sorted, the syntax A.merge(B) should merge all elements of $B$ into $A$ so that $A$ remains sorted and $B$ becomes empty. Your implementation must accomplish the merge by relinking existing nodes; you are not to create any new nodes.

```python
class _DoublyLinkedBase:

    class _Node:
        __slots__ = '_element', '_prev', '_next'

        def __init__(self, element, prev, next):
            self._element = element
            self._prev = prev
            self._next = next

    def __init__(self):
        self._header = self._Node(None, None, None)
        self._trailer = self._Node(None, None, None)
        self._header._next = self._trailer
        self._trailer._prev = self._header
        self._size = 0

    def __len__(self):
        return self._size

    def is_empty(self):
        return self._size == 0

    def _insert_between(self, e, predecessor, successor):
        newest = self._Node(e, predecessor, successor)
        predecessor._next = newest
        successor._prev = newest
        self._size += 1
        return newest

    def _delete_node(self, node):
        predecessor = node._prev
        successor = node._next
        predecessor._next = successor
        successor._prev = predecessor
        self._size -= 1
        element = node._element
        node._prev = node._next = node._element = None
        return element

# Código copiado del libro páginas 282-284
class PositionalList(_DoublyLinkedBase):
    """A sequential container of elements allowing positional access."""

    #-------------------------- nested Position class --------------------------
    class Position:
        """An abstraction representing the location of a single element."""

        def __init__(self, container, node):
            """Constructor should not be invoked by user."""
            self._container = container
            self._node = node

        def element(self):
            """Return the element stored at this Position."""
            return self._node._element

        def __eq__(self, other):
            """Return True if other is a Position representing the same location."""
            return type(other) is type(self) and other._node is self._node

        def __ne__(self, other):
            """Return True if other does not represent the same location."""
            return not (self == other)

    #------------------------------- utility method -------------------------------
    def _validate(self, p):
        """Return position's node, or raise appropriate error if invalid."""
        if not isinstance(p, self.Position):
            raise TypeError('p must be proper Position type')
        if p._container is not self:
            raise ValueError('p does not belong to this container')
        if p._node._next is None:
            raise ValueError('p is no longer valid')
        return p._node

    #------------------------------- utility method -------------------------------
    def _make_position(self, node):
        """Return Position instance for given node (or None if sentinel)."""
        if node is self._header or node is self._trailer:
            return None
        else:
            return self.Position(self, node)

    #------------------------------- accessors -------------------------------
    def first(self):
        """Return the first Position in the list (or None if list is empty)."""
        return self._make_position(self._header._next)

    def last(self):
        """Return the last Position in the list (or None if list is empty)."""
        return self._make_position(self._trailer._prev)

    def before(self, p):
        """Return the Position just before Position p (or None if p is first)."""
        node = self._validate(p)
        return self._make_position(node._prev)

    def after(self, p):
        """Return the Position just after Position p (or None if p is last)."""
        node = self._validate(p)
        return self._make_position(node._next)

    def __iter__(self):
        """Generate a forward iteration of the elements of the list."""
        cursor = self.first()
        while cursor is not None:
            yield cursor.element()
            cursor = self.after(cursor)

    #------------------------------- mutators -------------------------------
    # override inherited version to return Position, rather than Node
    def _insert_between(self, e, predecessor, successor):
        """Add element between existing nodes and return new Position."""
        node = super()._insert_between(e, predecessor, successor)
        return self._make_position(node)

    def add_first(self, e):
        """Insert element e at the front of the list and return new Position."""
        return self._insert_between(e, self._header, self._header._next)

    def add_last(self, e):
        """Insert element e at the back of the list and return new Position."""
        return self._insert_between(e, self._trailer._prev, self._trailer)

    def add_before(self, p, e):
        """Insert element e into list before Position p and return new Position."""
        original = self._validate(p)
        return self._insert_between(e, original._prev, original)

    def add_after(self, p, e):
        """Insert element e into list after Position p and return new Position."""
        original = self._validate(p)
        return self._insert_between(e, original, original._next)

    def delete(self, p):
        """Remove and return the element at Position p."""
        original = self._validate(p)
        return self._delete_node(original)

    def replace(self, p, e):
        """Replace the element at Position p with e.

        Return the element formerly at Position p.
        """
        original = self._validate(p)
        old_value = original._element
        original._element = e
        return old_value

    def merge(self, other):
        a = self._header._next
        b = other._header._next

        while b is not other._trailer:
            siguienteB = b._next

            while a is not self._trailer and a._element <= b._element:
                a = a._next
            antes = a._prev
            antes._next = b
            b._prev = antes
            b._next = a
            a._prev = b
            self._size = self._size + 1
            b = siguienteB
        other._header._next = other._trailer
        other._trailer._prev = other._header
        other._size = 0
```

```python
A = PositionalList()
B = PositionalList()

for x in [1,2,8,9,12]:
    A.add_last(x)
for y in [17,21,24]:
    B.add_last(y)
print("Lista A original: ", list(A))
print("Lista original B: ",list(B), "len de B:", len(B))
A.merge(B)
print("A:", list(A))
print("B:", list(B), "len de B:", len(B))
```
